```python
# 7개의 숫자로 100을 만들 수 있는 가짓수를 구하는 방법

from functools import lru_cache

@lru_cache(None)

def count_ways(remaining, choices_left) :
    if choices_left == 1 :
        return 1
    total = 0
    for i in range(remaining + 1) :
        total += count_ways(remaining - i, choices_left - 1)
    return total

print(count_ways(100, 7))
```

```python
from functools import lru_cache

@lru_cache(None)
def count_patterns(n):
    # 도시가 1개 남으면 그 도시가 우승하며 시나리오는 종료됩니다.
    if n == 1:
        return 1
    
    # 1. 과반수 득표자가 바로 나오는 경우 (1가지 패턴)
    total = 1
    
    # 2. 최하위가 1명(유일)인 경우
    # 남은 n개 도시 중 탈락할 1명을 고르는 방법은 n가지입니다.
    # 그 후 (n-1)개 도시에서의 패턴수를 곱합니다.
    total +=  count_patterns(n - 1)
    
    # 3. 최하위가 k명(2명 이상) 동률인 경우
    # n개 도시 중 k명이 동률 꼴찌가 되는 조합은 nCk입니다.
    # 그 k명이 투표해서 '1명을 제거'하는 과정의 패턴수는 count_patterns(k)와 같습니다.
    # 1명이 제거된 후 다시 (n-1)개 도시에서 투표를 이어갑니다.
    for k in range(2, n):
        import math
        combinations = math.comb(n, k)
        # (k명 중 1명 제거하는 패턴수) * (남은 n-1명으로 우승자 가리는 패턴수)
        total +=  count_patterns(k) * count_patterns(n - 1)
        
    return total

# 최종 결과 출력 (7개 도시)
n = 7
print(f"{n}개의 도시일 때 전체 패턴의 가짓수: {count_patterns(n)}")
```

```python
import math
from functools import lru_cache

@lru_cache(None)
def solve_voting(n):
    # 도시가 1개 남으면 이미 개최지가 결정된 상태 (종료)
    if n == 1:
        return 1
    
    # 1. 누군가 과반수로 즉시 당선되는 경우
    # 7개 도시 중 어느 도시가 당선될지 결정하는 가짓수
    count = n
    
    # 2. 꼴찌를 정해서 제거하는 과정
    # k명(1명~n-1명)이 꼴찌 후보가 되는 경우를 모두 따짐
    for k in range(1, n):
        # n명 중 k명의 꼴찌 후보를 선택하는 조합
        ways_to_choose_losers = math.comb(n, k)
        
        if k == 1:
            # 꼴찌가 1명이면 바로 제거하고 n-1명으로 진행
            count += ways_to_choose_losers * solve_voting(n - 1)
        else:
            # 꼴찌가 k명(여러 명)이면, 그 k명 중 한 명을 제거하는 
            # '소규모 투표 패턴수'를 먼저 구한 뒤(solve_voting(k)),
            # 제거된 후 남은 n-1명으로 진행
            count += ways_to_choose_losers * solve_voting(k) * solve_voting(n - 1)
            
    return count

print(f"7개 도시의 최종 패턴수: {solve_voting(7):,}")
```

```python
import math
from functools import lru_cache

@lru_cache(None)
def f(n):
    if n == 1:
        return 1
    
    # 1. 과반수 득표로 종료되는 패턴 (1가지)
    # (어떤 도시가 우승하느냐는 마지막에 n을 곱해서 처리하거나 
    # 혹은 패턴 자체의 '흐름'만 본다면 1로 시작합니다.)
    count = 1 
    
    # 2. 꼴찌 후보 k명을 선택하여 제거 프로세스 진행
    for k in range(1, n):
        # n명 중 k명을 꼴찌 후보로 선택하는 가짓수
        comb = math.comb(n, k)
        
        if k == 1:
            # 단독 꼴찌 제거 -> 남은 n-1명으로 계속
            count += comb * f(n - 1)
        else:
            # 동률 꼴찌 k명 발생 -> 그들 중 1명을 제거하는 과정(f(k))
            # 제거된 후 남은 n-1명으로 계속
            # 여기서 f(k)는 'k명 중 1명을 떨어뜨리는 패턴의 수'로 작동해야 합니다.
            count += comb * f(k) * f(n - 1)
            
    return count

# 만약 f(7)이 정답과 다르다면, '어느 도시가 우승하는가'의 n을 
# 재귀 내부가 아니라 최종 결과에 한 번만 고려해야 할 수도 있습니다.
# 문제의 '패턴' 정의에 따라 n=7일 때 f(7) * 7 혹은 f(7) 자체가 답이 됩니다.

print(f"7개 도시의 최종 패턴수: {f(7):,}")
```

```python
from itertools import product

def simulate(votes):
    """
    votes: tuple of 7 integers (각 도시의 득표수)
    return: 최종 개최 도시 인덱스 (0~6)
    """
    cities = list(range(7))
    votes = list(votes)

    while len(cities) > 1:
        total = sum(votes[c] for c in cities)
        majority = total // 2 + 1

        # 과반 확인
        for c in cities:
            if votes[c] >= majority:
                return c

        # 최하위 찾기
        min_vote = min(votes[c] for c in cities)
        losers = [c for c in cities if votes[c] == min_vote]

        if len(losers) == 1:
            # 꼴찌 하나 제거
            cities.remove(losers[0])
        else:
            # 동률이면 그들끼리 투표 -> 최하위 하나 제거
            # 여기서는 단순히 인덱스가 가장 작은 도시 제거 (패턴 수만 세기 위함)
            cities.remove(min(losers))

    return cities[0]


def count_patterns(max_votes=3):
    """
    max_votes: 각 도시가 받을 수 있는 최대 표 수 (탐색 범위 제한)
    """
    results = set()
    for votes in product(range(1, max_votes+1), repeat=7):
        if len(set(votes)) == 1:  # 모든 도시가 같은 표 제외
            continue
        winner = simulate(votes)
        results.add((votes, winner))
    return results


if __name__ == "__main__":
    patterns = count_patterns(max_votes=3)
    print("패턴 수:", len(patterns))
    # 예시로 몇 개 출력
    for i, (votes, winner) in enumerate(list(patterns)[:10]):
        print(f"{i+1}. 투표분포={votes}, 최종개최지=도시{winner+1}")
```
