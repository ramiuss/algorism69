```python
from itertools import permutations

M, N = 4, 3

seq = []
for a in permutations(list(range(2, M + 1)), M - 1) :
    seq.append([1] + list(a))

print(seq[0])
print(seq[1])
print(seq[2])
print(seq[3])
print(seq[-2])
print(seq[-1])
```

```python
log = []
log.append(seq)

print(len(log))
print(log[0][1])
print(log[0][-1])
```

```python
for i in range(0, N) :
    new_seq = []
    for a in log[0] :
        print(a, a[1])
        for j in range(1, M) :
            print(j, a[j], j + 1)
            if a[j] == j + 1 :
                new_seq.append(list(reversed(a[0 : (j + 1)])) + a[( j + 1) :])
                print(a[0 : (j + 1)] , a[( j + 1) :])
    log.append(new_seq)
    print(log, len(log))

print(len(log[N]))
```
