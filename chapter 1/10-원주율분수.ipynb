```python
from functools import lru_cache

@lru_cache(maxsize=None)
def pi(n) :
    result = []
    for i in range (3, 100) :
        for j in range (1, 100) :
            if 3.14 < i/ j < 3.15 :
                result.append((i, j))
                if len(result) == 1 :
                    break
    return result      

print(pi(2))
```
